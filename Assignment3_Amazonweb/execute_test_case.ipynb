```python
import os
import sys
sys.path.insert(0, r"c:\Users\kbhanushali1.ext\amazon-homepage-notes")
from playwright.sync_api import sync_playwright
from pages.home_page import HomePage
import json

root = r"c:\Users\kbhanushali1.ext\amazon-homepage-notes"
with open(os.path.join(root, "amazon_homepage_locators.json"), encoding="utf-8") as handle:
    locators = json.load(handle)

with sync_playwright() as p:
    browser = p.chromium.launch(headless=True)
    context = browser.new_context(viewport={"width": 1440, "height": 1200})
    page = context.new_page()
    home = HomePage(page, locators, "https://www.amazon.com")
    home.go_to()
    home.wait_for_load()
    print("title:", page.title())
    print("search visible:", home.get_locator("search_input").is_visible())
    context.close()
    browser.close()
```
